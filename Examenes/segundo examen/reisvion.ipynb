```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def blur_2(img,mask,blur):
    fondo=cv2.GaussianBlur(img,(51,51),0)
    fin=(np.where(mask[:,:,None]>0,img,fondo))
    return fin
imagen=cv2.imread('./rosa.png')
gray = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
_,mascara=cv2.threshold(gray,100,125,cv2.THRESH_BINARY)
mascara=cv2.bitwise_not(mascara)
mascara=cv2.GaussianBlur(mascara,(21,21),0)
blur=25
resultado=blur_2(imagen,mascara,blur)

plt.imshow(cv2.cvtColor(resultado,cv2.COLOR_BGR2RGB))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


imagen=cv2.imread('./elementos.jpg')
gray = cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)
_,binario=cv2.threshold(gray,128,255,cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
borde,_=cv2.findContours(binario,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
cont=len(borde)

plt.title(cont)
plt.imshow(binario)
```
